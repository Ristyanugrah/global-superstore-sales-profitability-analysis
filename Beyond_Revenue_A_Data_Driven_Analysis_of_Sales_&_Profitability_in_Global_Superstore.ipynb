```python
## 3.9 Deteksi Outlier

###Outlier diperiksa untuk mengetahui apakah terdapat nilai yang jauh berbeda dari sebagian besar observasi.

##Outlier tidak langsung dihapus karena nilai ekstrem pada data transaksi dapat merupakan transaksi bisnis yang valid. Oleh karena itu, hasil deteksi akan digunakan untuk investigasi lebih lanjut.
Q1_sales = df['Sales'].quantile(0.25)
Q3_sales = df['Sales'].quantile(0.75)

IQR_sales = Q3_sales - Q1_sales

lower_sales = Q1_sales - 1.5 * IQR_sales
upper_sales = Q3_sales + 1.5 * IQR_sales

outlier_sales = df[
    (df['Sales'] < lower_sales) |
    (df['Sales'] > upper_sales)
]

print("Q1 Sales :", Q1_sales)
print("Q3 Sales :", Q3_sales)
print("IQR Sales:", IQR_sales)

print(
    "Jumlah outlier Sales:",
    len(outlier_sales)
)

print(
    f"Persentase outlier Sales: {len(outlier_sales)/len(df)*100:.2f}%"
)
```

```python
##Outlier Profit
Q1_profit = df['Profit'].quantile(0.25)
Q3_profit = df['Profit'].quantile(0.75)

IQR_profit = Q3_profit - Q1_profit

lower_profit = Q1_profit - 1.5 * IQR_profit
upper_profit = Q3_profit + 1.5 * IQR_profit

outlier_profit = df[
    (df['Profit'] < lower_profit) |
    (df['Profit'] > upper_profit)
]

print("Q1 Profit :", Q1_profit)
print("Q3 Profit :", Q3_profit)
print("IQR Profit:", IQR_profit)

print(
    "Jumlah outlier Profit:",
    len(outlier_profit)
)

print(
    f"Persentase outlier Profit: {len(outlier_profit)/len(df)*100:.2f}%"
)
```

Variabel hasil transformasi akan digunakan pada tahap Exploratory Data Analysis (EDA), dashboard Power BI, serta business case analysis.

```python
# ============================================
# 4.1 PERSIAPAN DATASET
# ============================================

df['Order.Date'] = pd.to_datetime(df['Order.Date'])
df['Ship.Date'] = pd.to_datetime(df['Ship.Date'])

print("Order.Date :", df['Order.Date'].dtype)
print("Ship.Date  :", df['Ship.Date'].dtype)
```

```python
## 4.2 Shipping Days

##Shipping Days menunjukkan jumlah hari antara tanggal order dan tanggal pengiriman.

##Variabel ini digunakan untuk menganalisis operational performance dan membandingkan performa berdasarkan metode pengiriman.
df['Shipping.Days'] = (
    df['Ship.Date'] - df['Order.Date']
).dt.days

df[['Order.Date', 'Ship.Date', 'Shipping.Days']].head()
```

```python
print("Minimum Shipping Days :", df['Shipping.Days'].min())
print("Maximum Shipping Days :", df['Shipping.Days'].max())
print("Average Shipping Days :", round(df['Shipping.Days'].mean(), 2))
```

## 4.3 Profit Margin

Profit Margin digunakan untuk mengukur persentase profit yang dihasilkan dari setiap nilai sales.

Rumus:

Profit Margin = Profit / Sales

Metrik ini digunakan untuk membandingkan profitability antar kategori, produk, market, segment, dan discount level.

```python
df['Profit.Margin'] = np.where(
    df['Sales'] != 0,
    df['Profit'] / df['Sales'],
    0
)

df[['Sales', 'Profit', 'Profit.Margin']].head()
```

```python
print(
    "Rata-rata Profit Margin:",
    f"{df['Profit.Margin'].mean():.2%}"
)
```

## 4.4 Ekstraksi Variabel Waktu

Variabel waktu diturunkan dari Order.Date untuk mendukung analisis tren tahunan, bulanan, dan year-over-year.

```python
# Tahun
df['Order.Year'] = df['Order.Date'].dt.year

# Nomor bulan
df['Order.Month'] = df['Order.Date'].dt.month

# Nama bulan
df['Order.Month.Name'] = df['Order.Date'].dt.month_name()

# Tahun-Bulan
df['Order.Year.Month'] = df['Order.Date'].dt.to_period('M').astype(str)

df[
    [
        'Order.Date',
        'Order.Year',
        'Order.Month',
        'Order.Month.Name',
        'Order.Year.Month'
    ]
].head()
```

## 4.5 Discount Band

Discount dikelompokkan ke dalam beberapa rentang untuk mempermudah analisis hubungan antara tingkat discount dan profitability.

```python
def buat_discount_band(discount):
    if discount == 0:
        return '0%'
    elif discount <= 0.10:
        return '1-10%'
    elif discount <= 0.20:
        return '11-20%'
    elif discount <= 0.30:
        return '21-30%'
    elif discount <= 0.40:
        return '31-40%'
    else:
        return '>40%'

df['Discount.Band'] = df['Discount'].apply(buat_discount_band)

df['Discount.Band'].value_counts().sort_index()
```

```python
## 4.6 Profitability Status

#Setiap transaksi diklasifikasikan berdasarkan nilai Profit untuk membedakan transaksi yang menghasilkan profit dan transaksi yang mengalami loss.
df['Profitability.Status'] = np.where(
    df['Profit'] >= 0,
    'Profitable',
    'Loss'
)

df['Profitability.Status'].value_counts()
```

```python
profitability_distribution = (
    df['Profitability.Status']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

profitability_distribution
```

```python
## 4.7 Order-Level Performance

##Data transaksi diagregasi pada tingkat Order untuk mengetahui total sales, profit, quantity, discount, dan shipping cost pada setiap order.
order_analysis = (
    df.groupby('Order.ID')
      .agg(
          Order_Date=('Order.Date', 'min'),
          Customer_ID=('Customer.ID', 'first'),
          Sales=('Sales', 'sum'),
          Profit=('Profit', 'sum'),
          Quantity=('Quantity', 'sum'),
          Discount=('Discount', 'mean'),
          Shipping_Cost=('Shipping.Cost', 'sum'),
          Shipping_Days=('Shipping.Days', 'mean')
      )
      .reset_index()
)

order_analysis['Profit.Margin'] = np.where(
    order_analysis['Sales'] != 0,
    order_analysis['Profit'] / order_analysis['Sales'],
    0
)

order_analysis.head()
```

```python
## 4.9 Product-Level Performance

##Data diagregasi pada tingkat produk untuk mengidentifikasi produk dengan kontribusi sales dan profitability tertinggi maupun terendah.
product_analysis = (
    df.groupby(
        ['Product.ID', 'Product.Name', 'Category', 'Sub.Category']
    )
    .agg(
        Orders=('Order.ID', 'nunique'),
        Quantity=('Quantity', 'sum'),
        Sales=('Sales', 'sum'),
        Profit=('Profit', 'sum'),
        Avg_Discount=('Discount', 'mean')
    )
    .reset_index()
)

product_analysis['Profit.Margin'] = np.where(
    product_analysis['Sales'] != 0,
    product_analysis['Profit'] / product_analysis['Sales'],
    0
)

product_analysis.head()
```

```python
sales_median = product_analysis['Sales'].median()
profit_median = product_analysis['Profit'].median()

product_analysis['Performance.Group'] = np.select(
    [
        (product_analysis['Sales'] >= sales_median) &
        (product_analysis['Profit'] >= profit_median),

        (product_analysis['Sales'] >= sales_median) &
        (product_analysis['Profit'] < profit_median),

        (product_analysis['Sales'] < sales_median) &
        (product_analysis['Profit'] >= profit_median)
    ],
    [
        'High Sales - High Profit',
        'High Sales - Low Profit',
        'Low Sales - High Profit'
    ],
    default='Low Sales - Low Profit'
)

product_analysis['Performance.Group'].value_counts()
```

# 5. BUSINESS CASE ANALYSIS

## Investigating Profit Leakage

### Business Question
Apakah sales yang lebih tinggi selalu menghasilkan profitability yang lebih baik?

### Objective
Mengidentifikasi area dengan sales tinggi tetapi profit rendah atau negatif, kemudian menginvestigasi faktor yang berkaitan dengan kondisi tersebut.
